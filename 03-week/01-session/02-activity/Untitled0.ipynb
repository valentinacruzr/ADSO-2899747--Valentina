```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Cargar el archivo CSV
df = pd.read_csv("academic Stress level - maintainance 1.csv")

# Limpiar nombres de columnas (remover espacios extra)
df.columns = df.columns.str.strip()

# --- Gráfico 1: Distribución del nivel de estrés académico ---
plt.figure(figsize=(10, 6))
stress_counts = df['Rate your academic stress index'].value_counts().sort_index()
plt.bar(stress_counts.index, stress_counts.values, color='steelblue', alpha=0.7)
plt.title('Distribución del Índice de Estrés Académico', fontsize=16, fontweight='bold')
plt.xlabel('Nivel de Estrés (1-5)', fontsize=12)
plt.ylabel('Número de Estudiantes', fontsize=12)
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(stress_counts.values):
    plt.text(stress_counts.index[i], v + 0.5, str(v), ha='center', fontweight='bold')
plt.show()



```

```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt

# Cargar el archivo
df = pd.read_csv("Most Popular Programming Languages.csv")

df.head()

# --- Gráfico 1: Popularidad promedio de cada lenguaje ---
df.mean(numeric_only=True).sort_values(ascending=False).plot(
    kind="bar", figsize=(12,6), color="skyblue"
)
plt.title("Popularidad Promedio de Lenguajes de Programación")
plt.xlabel("Lenguaje")
plt.ylabel("Popularidad Promedio (%)")
plt.xticks(rotation=45, ha="right")
plt.show()

```

```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt

# Cargar archivo
df = pd.read_csv("covid_19_data.csv")

df.head()

# --- Gráfico 1: Top 10 países con más muertes ---
df.groupby("Region")["Deaths"].sum().sort_values(ascending=False).head(10).plot(
    kind="bar", figsize=(12,6), color="red"
)
plt.title("Top 10 países con más muertes")
plt.xlabel("País")
plt.ylabel("Muertes")
plt.xticks(rotation=45, ha="right")
plt.show()
```
