```python
# Subir archivo desde tu computadora
from google.colab import files
uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Cargar el archivo CSV
df = pd.read_csv("academic Stress level - maintainance 1.csv")

# Limpiar nombres de columnas (remover espacios extra)
df.columns = df.columns.str.strip()

# --- Gráfico 1: Distribución del nivel de estrés académico ---
plt.figure(figsize=(10, 6))
stress_counts = df['Rate your academic stress index'].value_counts().sort_index()
plt.bar(stress_counts.index, stress_counts.values, color='steelblue', alpha=0.7)
plt.title('Distribución del Índice de Estrés Académico', fontsize=16, fontweight='bold')
plt.xlabel('Nivel de Estrés (1-5)', fontsize=12)
plt.ylabel('Número de Estudiantes', fontsize=12)
plt.grid(axis='y', alpha=0.3)
for i, v in enumerate(stress_counts.values):
    plt.text(stress_counts.index[i], v + 0.5, str(v), ha='center', fontweight='bold')
plt.show()

# --- Gráfico 2: Estrés por etapa académica ---
plt.figure(figsize=(12, 6))
stage_stress = df.groupby('Your Academic Stage')['Rate your academic stress index'].mean().sort_values(ascending=False)
bars = plt.bar(range(len(stage_stress)), stage_stress.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1'])
plt.title('Nivel Promedio de Estrés por Etapa Académica', fontsize=16, fontweight='bold')
plt.xlabel('Etapa Académica', fontsize=12)
plt.ylabel('Estrés Promedio (1-5)', fontsize=12)
plt.xticks(range(len(stage_stress)), stage_stress.index, rotation=0)
plt.grid(axis='y', alpha=0.3)
# Agregar valores sobre las barras
for bar, value in zip(bars, stage_stress.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05,
             f'{value:.2f}', ha='center', fontweight='bold')
plt.show()

# --- Gráfico 3: Estrategias de afrontamiento más utilizadas ---
plt.figure(figsize=(14, 8))
coping_counts = df['What coping strategy you use as a student?'].value_counts()
colors = ['#FF9999', '#66B2FF', '#99FF99', '#FFCC99']
plt.pie(coping_counts.values, labels=coping_counts.index, autopct='%1.1f%%',
        colors=colors, startangle=90)
plt.title('Estrategias de Afrontamiento Utilizadas por los Estudiantes',
          fontsize=16, fontweight='bold')
plt.axis('equal')
plt.show()

# --- Gráfico 4: Relación entre presión académica en casa y estrés ---
plt.figure(figsize=(10, 6))
home_pressure_stress = df.groupby('Academic pressure from your home')['Rate your academic stress index'].mean()
plt.plot(home_pressure_stress.index, home_pressure_stress.values,
         marker='o', linewidth=3, markersize=8, color='red')
plt.title('Relación entre Presión Académica en Casa y Nivel de Estrés',
          fontsize=16, fontweight='bold')
plt.xlabel('Presión Académica en Casa (1-5)', fontsize=12)
plt.ylabel('Estrés Promedio (1-5)', fontsize=12)
plt.grid(True, alpha=0.3)
plt.show()

# --- Gráfico 5: Impacto del ambiente de estudio en el estrés ---
plt.figure(figsize=(12, 6))
env_stress = df.groupby('Study Environment')['Rate your academic stress index'].mean().sort_values(ascending=False)
bars = plt.bar(range(len(env_stress)), env_stress.values,
               color=['#FF6B6B', '#FFA726', '#66BB6A'])
plt.title('Nivel de Estrés según Ambiente de Estudio', fontsize=16, fontweight='bold')
plt.xlabel('Ambiente de Estudio', fontsize=12)
plt.ylabel('Estrés Promedio (1-5)', fontsize=12)
plt.xticks(range(len(env_stress)), env_stress.index, rotation=0)
plt.grid(axis='y', alpha=0.3)
# Agregar valores sobre las barras
for bar, value in zip(bars, env_stress.values):
    plt.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.05,
             f'{value:.2f}', ha='center', fontweight='bold')
plt.show()

```

```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt

# Cargar el archivo
df = pd.read_csv("Most Popular Programming Languages.csv")

df.head()

# --- Gráfico 1: Popularidad promedio de cada lenguaje ---
df.mean(numeric_only=True).sort_values(ascending=False).plot(
    kind="bar", figsize=(12,6), color="skyblue"
)
plt.title("Popularidad Promedio de Lenguajes de Programación")
plt.xlabel("Lenguaje")
plt.ylabel("Popularidad Promedio (%)")
plt.xticks(rotation=45, ha="right")
plt.show()

# --- Gráfico 2: Popularidad en el primer mes registrado ---
df.iloc[0, 1:].sort_values(ascending=False).plot(
    kind="bar", figsize=(12,6), color="lightgreen"
)
plt.title(f"Popularidad de Lenguajes en {df['Month'].iloc[0]}")
plt.xlabel("Lenguaje")
plt.ylabel("Popularidad (%)")
plt.xticks(rotation=45, ha="right")
plt.show()

# --- Gráfico 3: Popularidad en el último mes registrado ---
df.iloc[-1, 1:].sort_values(ascending=False).plot(
    kind="bar", figsize=(12,6), color="salmon"
)
plt.title(f"Popularidad de Lenguajes en {df['Month'].iloc[-1]}")
plt.xlabel("Lenguaje")
plt.ylabel("Popularidad (%)")
plt.xticks(rotation=45, ha="right")
plt.show()
```

```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import matplotlib.pyplot as plt

# Cargar archivo
df = pd.read_csv("covid_19_data.csv")

df.head()

# --- Gráfico 1: Top 10 países con más casos confirmados ---
df.groupby("Region")["Confirmed"].sum().sort_values(ascending=False).head(10).plot(
    kind="bar", figsize=(12,6), color="orange"
)
plt.title("Top 10 países con más casos confirmados")
plt.xlabel("País")
plt.ylabel("Casos Confirmados")
plt.xticks(rotation=45, ha="right")
plt.show()

# --- Gráfico 2: Top 10 países con más muertes ---
df.groupby("Region")["Deaths"].sum().sort_values(ascending=False).head(10).plot(
    kind="bar", figsize=(12,6), color="red"
)
plt.title("Top 10 países con más muertes")
plt.xlabel("País")
plt.ylabel("Muertes")
plt.xticks(rotation=45, ha="right")
plt.show()

# --- Gráfico 3: Top 10 países con más recuperados ---
df.groupby("Region")["Recovered"].sum().sort_values(ascending=False).head(10).plot(
    kind="bar", figsize=(12,6), color="green"
)
plt.title("Top 10 países con más recuperados")
plt.xlabel("País")
plt.ylabel("Recuperados")
plt.xticks(rotation=45, ha="right")
plt.show()
```
